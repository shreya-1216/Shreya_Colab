```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 170, 160, 200, 230]
}
df = pd.DataFrame(data)
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Department": ["Marketing", "IT", "HR", "Operations", "Research"],
    "Budget": [30, 25, 15, 20, 10]
}
df = pd.DataFrame(data)
plt.pie(
    df["Budget"],
    labels=df["Department"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Company Budget Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt
class_A = [65, 70, 72, 75, 78, 80, 82, 85]
class_B = [50, 55, 60, 65, 70, 75, 80, 85]
class_C = [75, 78, 80, 82, 84, 86, 88, 90]
plt.boxplot(
    [class_A, class_B, class_C],
    tick_labels=["Class A", "Class B", "Class C"]
)
plt.title("Marks Distribution by Class")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
visitors = [1000, 1400, 1800, 1600, 2200, 2700]
plt.fill_between(months, visitors, alpha=0.5)
plt.plot(months, visitors, marker="o")
plt.title("Website Visitors")
plt.xlabel("Month")
plt.ylabel("Visitors")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
quarters = ["Q1", "Q2", "Q3", "Q4"]
product_A = [120, 150, 170, 200]
product_B = [100, 130, 160, 190]
x = np.arange(len(quarters))
width = 0.35
plt.bar(x - width/2, product_A, width, label="Product A")
plt.bar(x + width/2, product_B, width, label="Product B")
plt.xticks(x, quarters)
plt.title("Quarterly Product Sales")
plt.xlabel("Quarter")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
departments = ["CS", "IT", "Business", "Arts"]
male = [60, 50, 40, 30]
female = [40, 45, 55, 70]
plt.bar(departments, male, label="Male")
plt.bar(departments, female, bottom=male, label="Female")
plt.title("Students by Department")
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_A = [100, 120, 140, 160, 180, 210]
product_B = [90, 110, 150, 145, 170, 190]
plt.plot(months, product_A, marker="o", label="Product A")
plt.plot(months, product_B, marker="o", label="Product B")
plt.title("Monthly Product Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
data = np.array([
    [20, 40, 60],
    [25, 45, 65],
    [30, 50, 70],
    [35, 55, 75],
    [40, 70, 100]
])
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
times = ["Morning", "Afternoon", "Evening"]
plt.imshow(data)
plt.xticks(range(len(times)), times)
plt.yticks(range(len(days)), days)
plt.colorbar(label="Customers")
plt.title("Customer Visits Heatmap")
plt.show()
```

```python
import matplotlib.pyplot as plt
hours = [1, 2, 3, 4, 5, 6, 7]
calls = [5, 8, 4, 10, 7, 12, 9]
plt.stem(hours, calls)
plt.title("Calls Received Per Hour")
plt.xlabel("Hour")
plt.ylabel("Number of Calls")
plt.show()
```

```python
import matplotlib.pyplot as plt
revenue = [100, 150, 80, 200, 170]
profit = [20, 35, 15, 50, 40]
employees = [100, 200, 70, 300, 220]
plt.scatter(
    revenue,
    profit,
    s=employees,
    alpha=0.5
)
plt.title("Company Performance")
plt.xlabel("Revenue")
plt.ylabel("Profit")
plt.show()
```

```python
import matplotlib.pyplot as plt
experiments = ["A", "B", "C", "D"]
average = [50, 55, 53, 60]
error = [2, 3, 1.5, 2.5]
plt.errorbar(
    experiments,
    average,
    yerr=error,
    fmt="o",
    capsize=5
)
plt.title("Experimental Results")
plt.xlabel("Experiment")
plt.ylabel("Average Measurement")
plt.show()
```
